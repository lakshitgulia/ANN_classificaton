

Prediction if the persun will chutn the job or not from input data using the model

```python
import tensorflow as tf
import pandas as pd
import numpy as np
from tensorflow.keras.models import load_model 
import pickle
```

```python
### LOAD THE ANN TRAINED MODEL , SACLER PICKLE FILE AND ONE HOT PICKLE FILE
model=load_model('model.h5')

## load the encoders and scalers

with open('lable_encoder_gender.pkl','rb') as file:
    lable_encoder_gender = pickle.load(file)

with open('onehot_encoder_geo.pkl','rb') as file:
    onehot_encoder_geo = pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
## Input data 

input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
## encode the input geography value

geo_encoded = onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()

geo_encode_df = pd.DataFrame(
    geo_encoded,
    columns=onehot_encoder_geo.get_feature_names_out(['Geography'])
)
geo_encode_df
```

```python

input_df=pd.DataFrame([input_data])
input_df
```

```python
## encode the input gender value

input_df['Gender']= lable_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
## Conatinate one hot encoded

input_df=pd.concat([input_df.drop("Geography",axis=1),geo_encode_df],axis=1)
input_df
```

```python
## Scaleing the input data

input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## Predict Churn

prediction=model.predict(input_scaled)
prediction
```

```python
prediction_probab= prediction[0][0]
prediction_probab
```

```python
if prediction_probab > 0.5:
    print('will churn')
else:
    print('will not churn')
        
```
